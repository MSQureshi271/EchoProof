## Project path

```python
PROJECT_ROOT = Path('/content/drive/MyDrive/PAI&AIES Project')  # edit if needed
ALLOW_PACKAGE_INSTALL = True
for required in ('IMPLEMENTATION_PLAN.md', 'pyproject.toml', 'src/echoproof/cli.py', 'docs/phase1_setup.md', 'scripts/p1_03_colab_setup.py'):
    if not (PROJECT_ROOT / required).is_file():
        raise FileNotFoundError(f'Sync {required} to Drive first')
print('Project:', PROJECT_ROOT)
```

## Run package setup and checks

This cell records Python/torch before and after installation, baseline imports, the module and console CLI, `pip check`, FFmpeg and MP3 decoder/encoder availability. A nonzero check is reported as `needs_attention`; read the saved diagnostics before retrying. If Colab requests a runtime restart after package installation, reconnect, remount Drive and run again.
```python
import importlib.util

script = PROJECT_ROOT / 'scripts/p1_03_colab_setup.py'
spec = importlib.util.spec_from_file_location('echoproof_p1_03_setup', script)
module = importlib.util.module_from_spec(spec)
spec.loader.exec_module(module)
REPORT, OUTPUT_DIR = module.setup(PROJECT_ROOT, allow_install=ALLOW_PACKAGE_INSTALL)
print('Status:', REPORT['status'])
print('Saved evidence:', OUTPUT_DIR)
print('Issues:', REPORT['issues'])
```
