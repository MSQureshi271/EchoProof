# R2-M2: BiGRU training-only diagnostics
Use a Colab T4 runtime. Sync the source, script and tests to the Drive project. The diagnostic reads 32 B-training recordings, trains disposable weights in memory, and saves only the evidence JSON.

```python
from google.colab import drive
drive.mount("/content/drive")
from pathlib import Path
import os, sys, subprocess, time
ROOT = Path("/content/drive/MyDrive/PAI&AIES Project")
SHARDS = ROOT / "data/processed/cohort_v1"
assert (ROOT / "scripts/r2_m2_bigru.py").exists() and SHARDS.exists()
os.chdir(ROOT)
os.environ["PYTHONPATH"] = str(ROOT / "src")
import torch
assert torch.cuda.is_available(), "Select a GPU runtime for the real-data diagnostic"
```

```python
subprocess.run([sys.executable, "-m", "pytest", "tests/test_r2_m2_diagnostic.py", "-q", "-p", "no:cacheprovider"], check=True)
```

```python
report = ROOT / "reports/r2" / ("r2_m2_" + time.strftime("%Y%m%dT%H%M%SZ", time.gmtime()) + ".json")
subprocess.run([sys.executable, "scripts/r2_m2_bigru.py", "--root", str(ROOT), "--shard-dir", str(SHARDS), "--report", str(report), "--device", "cuda"], check=True)
print("Evidence:", report)
```
